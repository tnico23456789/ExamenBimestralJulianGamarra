#Pregunta 1

```python
import pandas as pd
df1 = pd.read_csv('clientes.csv')
df2 = pd.read_csv('productos_.csv')
df3 = pd.read_csv('ventas.csv')
```

```python
ticketsunicos = df3['id_ticket'].nunique()
total_lineas_venta = len(df3)
productos_vendidos =df3['cantidad'].mean() * df3['cantidad'].count()
tabla1 = pd.DataFrame({
    'Metrica' : ['Total lineas de venta', 'cantidad tickets distintos', 'Total productos vendidos' ],
    'Resultado' : [total_lineas_venta, ticketsunicos, productos_vendidos ]
})
display(tabla1)
```

#Pregunta 2

```python
tabla2 = pd.crosstab(
    columns= df3['id_ticket'].drop_duplicates(), 
    index=  df3['sucursal'],
    margins=True,
    margins_name='Total'
)

display(tabla2)

print("La sucursal Central vendió mas tickets")
```

#Pregunta 3

```python
tabla3 = pd.crosstab(
    index = df3['metodo_pago'],
    columns= df3['id_ticket'].drop_duplicates(),
    margins=True,
    margins_name='Total'
)

tabla_3 = pd.crosstab(
    index = df3['metodo_pago'].notna(),
    columns= df3['id_ticket'].drop_duplicates(),
    margins=True,
    margins_name='Total'
)
tabla_3.index = ['Sin Datos', 'Con datos','Total']

display(tabla3)
display(tabla_3)
```

#Pregunta 4

```python
tabla4 = df2.groupby('categoria')['precio_lista'].agg(
    Cantidad_Productos='count',
    Precio_Promedio='mean',
    Precio_Minimo='min',
    Precio_Maximo='max'
)

display(tabla4)
```
